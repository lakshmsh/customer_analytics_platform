# 🎁 Advanced Hybrid Recommender System (Bonus 1)

## Architecture:
- **Item-Item Collaborative Filtering**: Computes cosine similarity over user-item log-interaction matrix.
- **Content-Based Filtering**: TF-IDF vectorization on product descriptions to recommend thematic relatives.
- **Hybrid Score Blending**: $\text{Score} = \alpha \cdot \text{CF} + (1-\alpha) \cdot \text{Content}$.
- **Churn Retention Incentives**: Generates personalized dynamic discounts (e.g. 15% VIP discount) for high-churn accounts.
