# 👥 Customer Segmentation & Persona Profiling

## Objectives:
- Apply unsupervised clustering (K-Means and Gaussian Mixture Models) on scaled RFM and basket dynamics.
- Identify optimal cluster count ($k=4$) via Silhouette analysis and Davies-Bouldin index.
- Map empirical clusters to business personas (*Champions*, *Potential Loyalists*, *At-Risk High Rollers*, *Hibernating*).

```python
import os, sys
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
os.chdir(project_root)
sys.path.insert(0, str(project_root))

with open('models/segmentation_metadata.json', 'r') as f:
    meta = json.load(f)

print('Clustering Quality:', meta['metrics'])
profiles_df = pd.DataFrame(meta['cluster_profiles'])
display(profiles_df)
```

```python
# Visualize Persona Cluster Sizes and Churn Rates
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(data=profiles_df, x='persona_name', y='size', palette='Blues_r', ax=axes[0])
axes[0].set_title('Customer Count by Persona', fontweight='bold')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=20)

sns.barplot(data=profiles_df, x='persona_name', y='churn_rate', palette='Reds_r', ax=axes[1])
axes[1].set_title('Churn Rate by Persona (%)', fontweight='bold')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=20)
plt.tight_layout()
plt.show()
```
